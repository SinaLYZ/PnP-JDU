```python
import scipy.io
import matplotlib.pyplot as plt
import numpy as np
import os

# Load kernels from different files
kernel_files = {
    'kernel_1': './kernel_1.mat',
    'kernel_2': './kernel_2.mat',
    'kernel_3': './kernel_3.mat',
    'kernel_5': './kernel_5.mat',
    'kernel_6': './kernel_6.mat'
}

# Dictionary to store kernels
kernels = {}

for key, file in kernel_files.items():
    data = scipy.io.loadmat(file)
    kernels[key] = data['kernel']

# Optional placeholder
kernels['kernel_4'] = np.zeros_like(next(iter(kernels.values())))

# Output directory
out_dir = './kernel_plots'
os.makedirs(out_dir, exist_ok=True)

for key in sorted(kernels.keys()):
    kernel = kernels[key]

    y = np.arange(kernel.shape[0])
    x = np.arange(kernel.shape[1])
    X, Y = np.meshgrid(x, y)

    fig = plt.figure(figsize=(10, 8))
    ax = fig.add_subplot(111, projection='3d')

    ax.plot_surface(
        X, Y, kernel,
        cmap='viridis',
        edgecolor='k',
        alpha=0.85
    )

    
    ax.set_xlabel("X")
    ax.set_ylabel("Y")
    ax.set_zlabel("Amplitude")
    ax.view_init(elev=30, azim=135)

    # Save figure
    save_path = os.path.join(out_dir, f"{key}_3d.png")
    plt.savefig(save_path, dpi=300, bbox_inches='tight', pad_inches=0.2)
    plt.close(fig)

print(f"Saved kernel plots to: {out_dir}")
```
